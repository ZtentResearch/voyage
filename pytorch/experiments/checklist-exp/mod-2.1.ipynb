```python
import torch
from torchviz import make_dot
```

```python
def p(whatever):
    print(whatever)
```

```python
x = torch.tensor(2.0, requires_grad=True)
w = torch.tensor(3.0, requires_grad=True)
b = torch.tensor(4.0, requires_grad=True)

y = x * w
z = y + b
loss = z ** 2
loss
```

```python
p(loss.grad_fn)
p(loss.grad_fn.next_functions)
p(loss.grad_fn.next_functions[0][0].next_functions)
```

```python
p(x.is_leaf)
p(y.is_leaf)
```

```python
graph = make_dot(loss, params={"x": x, "w": w, "b": b, "y": y, "z": z, "loss": loss})
graph.render('graph', format='svg')
```

```python
y.retain_grad()
z.retain_grad()
loss.retain_grad()
loss.backward()
```

```python
p(loss.grad)
p(z.grad)
p(y.grad)
p(x.grad)
p(w.grad)
p(b.grad)
```
